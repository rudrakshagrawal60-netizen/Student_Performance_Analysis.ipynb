```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as pit
```

```python
# create the line graph
# create a dataset
subjects = ["Python", "Statistics", "DBMS", "Machine Learning", "Data Visualization"]
marks = [85, 75, 90, 82, 88]

# display the data
print("Subjects:", subjects)
print("Marks:", marks)
```

```python
import matplotlib.pyplot as plt

# create the line graph
plt.plot(subjects, marks)
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Student Performance Analysis")
plt.xticks(rotation=30)
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Customization of the graph
plt.figure(figsize=(8, 5))  # width = 8, height = 5

plt.plot(
    subjects,
    marks,
    color='green',
    linestyle='--',
    linewidth=5,
    marker='s',                 # Square marker
    markersize=10,
    markerfacecolor='yellow',   # Yellow fill inside the marker
    markeredgecolor='red',      # Red border around the marker
    markeredgewidth=2,          # Thickness of the red border
    label='Student Marks'       # Legend mein dikhane ke liye label joda gaya hai
)

plt.title("Students Performance Analysis")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.grid(linestyle="--", alpha=0.8)  # alpha controls transparency (range 0-1)
plt.xticks(rotation=45)
plt.legend(loc='upper right')        # Display the legend in the top right
plt.show()
```

```python
import matplotlib.pyplot as plt

# Set figure size before plotting
plt.figure(figsize=(5, 4))

# Bar Plot
plt.bar(
    subjects,
    marks,
    color="pink",
    edgecolor="green",
    width=0.6
)
plt.title("Subject Mark Comparison")

# Label for x and y axis
plt.xlabel("Subjects")
plt.ylabel("Marks")

# Display the marks above the bar
for i, value in enumerate(marks):
    plt.text(i, value + 1, str(value), ha="center")

plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.ylim(0, 100)

# Display the graph
plt.show()
```

```python
# horizontal bar graph
plt.figure(figsize=(8,5))
plt.barh(subjects, marks, color="orange")
plt.title("Horizontal Bar Graph")
plt.xlabel("marks")
plt.ylabel("students")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Scatter plot
studyhours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [35, 40, 50, 55, 65, 70, 80, 90]

# Create the scatter plot
plt.scatter(
    studyhours,
    marks,
    s=150,                  # size = 150
    c="red",
    marker="o",             # Changed '0' (zero) to 'o' (letter o) for circle
    alpha=0.8,              # Added missing comma here
    edgecolor="black",
    label="Student Data"    # Added label so plt.legend() works without warning
)

# Title and labels
plt.title("Scatter plot for study hours", fontsize=16, fontweight="bold")
plt.xlabel("study hours")
plt.ylabel("Examination Mark")
plt.xticks(studyhours)
plt.yticks(range(0, 101, 10))

plt.xlim(0, 9)
plt.ylim(0, 100)

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Pie chart
attendance = ["Present", "Absent"]
count = [80, 20]

# Create the pie chart
plt.pie(
    count,
    labels=attendance,
    autopct='%1.1f%%',    # Changed comma (,) to period (.) in %1.1f%%
    startangle=90
)

plt.title("Student attendance distribution")  # Added quotation marks around the title
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Histogram
marks = [
    35, 40, 45, 48, 50,
    52, 55, 58, 60, 62,
    65, 68, 70, 72, 75,
    78, 80, 85, 90, 95
]

# Set figure size BEFORE creating the histogram
plt.figure(figsize=(5, 5))

# Create the histogram
plt.hist(
    marks,
    bins=5,
    color="lightgreen",
    edgecolor="black"
)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```
